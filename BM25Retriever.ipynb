```python
import os
from dotenv import load_dotenv
from llama_index.llms.google_genai import GoogleGenAI

load_dotenv()

api_key = os.environ.get("GOOGLE_API_KEY")

llm = GoogleGenAI(model="gemini-2.0-flash", api_key=api_key)
```

```python
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

embed_model = HuggingFaceEmbedding(model_name="BAAI/bge-small-en")
```

```python
from llama_index.core import Settings

Settings.llm = llm
Settings.embed_model = embed_model
```

```python
from llama_index.core import SimpleDirectoryReader

documents = SimpleDirectoryReader(input_dir="data/paul_graham").load_data()
```

```python
from llama_index.core.node_parser import SentenceSplitter

splitter = SentenceSplitter(chunk_size=512)

nodes = splitter.get_nodes_from_documents(documents)
```

```python
from llama_index.retrievers.bm25 import BM25Retriever
import Stemmer

bm_25_retriever = BM25Retriever.from_defaults(
    nodes=nodes,
    similarity_top_k=2,
    stemmer=Stemmer.Stemmer("english"),
    language='english'
)
```

```python
from llama_index.core.storage.docstore import SimpleDocumentStore

docstore = SimpleDocumentStore()
docstore.add_documents(nodes)
```

```python
from llama_index.core.response.notebook_utils import display_source_node

retrieved_nodes = bm_25_retriever.retrieve("What happened at Viaweb and Interleaf?")

for node in retrieved_nodes:
    display_source_node(node, source_length=5000)
```
